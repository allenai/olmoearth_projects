# Change Detection: Park Fire

This notebook computes per-pixel cosine distance between annual
OlmoEarth embeddings from 2023 and 2025 to detect
the Park Fire burn scar in Butte County, California.

**Prerequisites:** Run `PYTHONPATH=src python -m olmoearth_embeddings_tutorial.change_detection.download` first to download
embeddings and Sentinel-2 RGB for both periods.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt

# Allow imports from the tutorial src/ directory
sys.path.insert(0, str(Path.cwd().parent / "src"))

from olmoearth_embeddings_tutorial.change_detection.analyze import (
    compute_change,
    make_change_figure,
)
```

```python
BEFORE_DIR = Path("../data/change_detection/2023")
AFTER_DIR = Path("../data/change_detection/2025")

result = compute_change(BEFORE_DIR, AFTER_DIR)
```

```python
import numpy as np

valid_change = result.change[result.valid]
print(f"Median cosine distance: {np.nanmedian(valid_change):.4f}")
print(f"95th percentile:        {np.nanpercentile(valid_change, 95):.4f}")
print(f"99th percentile:        {np.nanpercentile(valid_change, 99):.4f}")
```

```python
fig = make_change_figure(result)
plt.show()
```
