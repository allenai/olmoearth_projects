# Few-Shot Land-Cover Segmentation

This notebook trains a logistic regression on just 60 labeled pixels
(20 per class) from Ca Mau, Vietnam and classifies every pixel in the
region into mangrove, water, or other.

**Prerequisites:** Run `PYTHONPATH=src python -m olmoearth_embeddings_tutorial.segmentation.download` first to download
embeddings, Sentinel-2 RGB, and WorldCover.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt

# Allow imports from the tutorial src/ directory
sys.path.insert(0, str(Path.cwd().parent / "src"))

from olmoearth_embeddings_tutorial.segmentation.analyze import (
    make_fewshot_figure,
    train_fewshot,
)
```

```python
DATA_DIR = Path("../data/segmentation/ca_mau")

result = train_fewshot(DATA_DIR, n_per_class=20)
```

```python
print(f"F1:   {result.f1:.3f}")
print(f"mIoU: {result.miou:.3f}")
```

```python
fig = make_fewshot_figure(result)
plt.show()
```
