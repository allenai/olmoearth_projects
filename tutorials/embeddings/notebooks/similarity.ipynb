# Similarity Search with OlmoEarth Embeddings

This notebook computes per-pixel cosine similarity to a query location and
visualizes the results as a heatmap and patch mosaic.

**Prerequisites:** Run `PYTHONPATH=src python -m olmoearth_embeddings_tutorial.similarity.download` first to download
embeddings and Sentinel-2 RGB for the Central Valley.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt

# Allow imports from the tutorial src/ directory
sys.path.insert(0, str(Path.cwd().parent / "src"))

from olmoearth_embeddings_tutorial.common.constants import (
    EMBEDDINGS_FILENAME,
    S2_RGB_FILENAME,
)
from olmoearth_embeddings_tutorial.similarity.analyze import (
    compute_similarity,
    make_heatmap_figure,
    make_mosaic_figure,
)
```

```python
EMBED_PATH = Path("../data/central_valley") / EMBEDDINGS_FILENAME
RGB_PATH = Path("../data/central_valley") / S2_RGB_FILENAME

result = compute_similarity(EMBED_PATH, RGB_PATH)
```

```python
fig = make_heatmap_figure(result)
plt.show()
```

```python
fig = make_mosaic_figure(result)
plt.show()
```
